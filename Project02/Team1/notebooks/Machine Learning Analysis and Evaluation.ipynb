```python
import os
import logging
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display, Markdown

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score,
    average_precision_score, recall_score, f1_score,
    precision_score, accuracy_score, roc_curve, precision_recall_curve,
    brier_score_loss
)
from sklearn.calibration import calibration_curve
from sklearn.model_selection import (
    RandomizedSearchCV, PredefinedSplit, GroupKFold, StratifiedKFold
)
from scipy.stats import chi2_contingency

import xgboost as xgb

# -----------------------------------------------------------------------------
# Configuration
# -----------------------------------------------------------------------------
logging.basicConfig(level=logging.INFO,
                    format="%(asctime)s - %(levelname)s - %(message)s")
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update({"font.size": 11, "figure.autolayout": True})

SEED = 42
FIG_DIR = "../figures"          # FIX (figures path)
os.makedirs(FIG_DIR, exist_ok=True)
np.random.seed(SEED)


# -----------------------------------------------------------------------------
# Helpers
# -----------------------------------------------------------------------------
def slug(name: str) -> str:
    return (name.lower()
            .replace(" ", "_").replace("(", "").replace(")", "")
            .replace("-", "_").replace("/", "_").replace("=", "")
            .replace(",", "").replace(".", ""))


def save_show(fig, path):
    fig.savefig(path, dpi=300, bbox_inches="tight")
    plt.show()
    plt.close(fig)
    logging.info(f"Saved: {path}")


def xgb_gain_with_names(model, feature_names):
    """XGBoost gain importance as {feature_name: gain}."""
    booster = model.get_booster()
    try:
        booster.feature_names = list(feature_names)
    except Exception:
        pass
    gain = booster.get_score(importance_type='gain')
    out = {}
    for k, v in gain.items():
        if (isinstance(k, str) and len(k) > 1
                and k[0] == 'f' and k[1:].isdigit()):
            idx = int(k[1:])
            name = feature_names[idx] if 0 <= idx < len(feature_names) else k
            out[name] = v
        else:
            out[k] = v
    return out


def wilson_ci(k, n, z=1.96):
    """Wilson score interval for a binomial proportion.

    Guarantees lo <= k/n <= hi after clamping to [0, 1].  The clamp is what
    stops `recall - recall_lo` (a `yerr` half-width) from ever being a
    tiny negative number due to floating-point drift.
    """
    if n == 0:
        return (np.nan, np.nan)
    p = k / n
    denom = 1 + z ** 2 / n
    centre = (p + z ** 2 / (2 * n)) / denom
    half = z * np.sqrt(p * (1 - p) / n + z ** 2 / (4 * n ** 2)) / denom
    lo = max(0.0, min(p, centre - half))
    hi = min(1.0, max(p, centre + half))
    return (float(lo), float(hi))


def bootstrap_ci(y_true, y_prob, metric_fn, n_boot=1000, alpha=0.05,
                 seed=SEED):
    rng = np.random.default_rng(seed)
    n = len(y_true)
    scores = []
    for _ in range(n_boot):
        idx = rng.integers(0, n, n)
        try:
            scores.append(metric_fn(y_true[idx], y_prob[idx]))
        except Exception:
            continue
    if not scores:
        return (np.nan, np.nan, np.nan)
    return (float(np.mean(scores)),
            float(np.quantile(scores, alpha / 2)),
            float(np.quantile(scores, 1 - alpha / 2)))


def pick_threshold(y_true, probs, target_recall=0.60):
    """
    Return the HIGHEST threshold that still achieves recall >= target_recall.
    """
    best = None
    for t in np.linspace(0.95, 0.05, 91):          # descending
        y_hat = (probs >= t).astype(int)
        rec = recall_score(y_true, y_hat, zero_division=0)
        if rec >= target_recall:
            prec = precision_score(y_true, y_hat, zero_division=0)
            f1v = f1_score(y_true, y_hat, zero_division=0)
            best = (float(t), rec, prec, f1v)
            break
    return best


# =============================================================================
# SECTION 1. DATA LOADING AND PATIENT-LEVEL SPLIT
# =============================================================================
logging.info("=== Section 1: Data loading and split ===")

DATA_PATH = "../data/diabetic_cleaned.csv"
df = pd.read_csv(DATA_PATH)

if 'target' not in df.columns:
    if 'readmitted' not in df.columns:
        raise ValueError("Need either a 'target' or a 'readmitted' column.")
    df['target'] = (df['readmitted'] == '<30').astype(int)
    logging.info(
        f"Target = (readmitted == '<30'): positive rate "
        f"{df['target'].mean():.3f} (n_pos={int(df['target'].sum())}, "
        f"n={len(df)})")
else:
    logging.info(f"Using existing target column: "
                 f"positive rate {df['target'].mean():.3f}")

TRAIN_PCT, VAL_PCT, TEST_PCT = 80.0, 10.0, 10.0
PATIENT_IDS_PATH = "../data/patient_ids_for_split.csv"

# ---- Attach patient_nbr if possible -----------------------------------------
if os.path.exists(PATIENT_IDS_PATH):
    patient_ids = pd.read_csv(PATIENT_IDS_PATH)
    if 'patient_nbr' in patient_ids.columns:
        if 'patient_nbr' not in df.columns:
            if len(patient_ids) == len(df):
                df['patient_nbr'] = patient_ids['patient_nbr'].values
                logging.info("Attached patient_nbr by row order.")
            else:
                logging.warning(
                    f"Length mismatch: ids={len(patient_ids)} df={len(df)}; "
                    "cannot attach patient_nbr.")
        else:
            logging.info("df already contains patient_nbr.")
    else:
        logging.warning("patient_ids_for_split.csv has no patient_nbr column.")

# ---- Use provided split if present -----------------------------------------
if os.path.exists(PATIENT_IDS_PATH):
    splits_df = pd.read_csv(PATIENT_IDS_PATH)
    if ('split' in splits_df.columns
            and 'patient_nbr' in splits_df.columns
            and 'patient_nbr' in df.columns):
        df = df.merge(
            splits_df[['patient_nbr', 'split']].drop_duplicates(),
            on='patient_nbr', how='left'
        )
        logging.info("Merged provided split assignments.")

if 'split' not in df.columns or df['split'].isna().any():
    if 'patient_nbr' in df.columns and df['patient_nbr'].notna().all():
        logging.warning(
            "No usable precomputed split found → performing PATIENT-GROUPED "
            "split (all encounters of a patient stay in the same split).")
        rng = np.random.default_rng(SEED)
        unique_pats = df['patient_nbr'].unique()
        rng.shuffle(unique_pats)
        n = len(unique_pats)
        n_tr = int(round(TRAIN_PCT / 100 * n))
        n_va = int(round(VAL_PCT / 100 * n))
        train_pats = set(unique_pats[:n_tr])
        val_pats = set(unique_pats[n_tr:n_tr + n_va])
        test_pats = set(unique_pats[n_tr + n_va:])
        df['split'] = np.where(
            df['patient_nbr'].isin(train_pats), 'train',
            np.where(df['patient_nbr'].isin(val_pats), 'val',
                     np.where(df['patient_nbr'].isin(test_pats), 'test',
                              'unassigned')))
    else:
        logging.warning(
            "!!! patient_nbr not available → falling back to an "
            "ENCOUNTER-LEVEL split. Encounters of the same patient may end "
            "up in different splits; treat metrics as optimistic !!!")
        rng = np.random.default_rng(SEED)
        idx = np.arange(len(df))
        rng.shuffle(idx)
        n = len(df)
        n_tr = int(round(TRAIN_PCT / 100 * n))
        n_va = int(round(VAL_PCT / 100 * n))
        split_col = np.array(['test'] * n, dtype=object)
        split_col[idx[:n_tr]] = 'train'
        split_col[idx[n_tr:n_tr + n_va]] = 'val'
        df['split'] = split_col

# ---- Derived columns on the full dataframe (so OOF fairness can use them) --
if all(c in df.columns for c in
       ['number_outpatient', 'number_emergency', 'number_inpatient']):
    df['total_visits'] = (df['number_outpatient']
                          + df['number_emergency']
                          + df['number_inpatient'])
    df['inpatient_ratio'] = df['number_inpatient'] / (df['total_visits'] + 1e-5)

train_df = df[df['split'] == 'train'].copy()
val_df = df[df['split'] == 'val'].copy()
test_df = df[df['split'] == 'test'].copy()

logging.info(f"Full: {df.shape} | Train: {len(train_df)} "
             f"| Val: {len(val_df)} | Test: {len(test_df)}")
logging.info(f"Target rate — train {train_df['target'].mean():.3f}, "
             f"val {val_df['target'].mean():.3f}, "
             f"test {test_df['target'].mean():.3f}")


# =============================================================================
# SECTION 2. Q4 — BASELINE MODELS (LR + DECISION TREE)
# =============================================================================
logging.info("=== Section 2 (Q4): Baseline models ===")

num_cols_base = [
    'time_in_hospital', 'num_lab_procedures', 'num_procedures',
    'num_medications', 'number_outpatient', 'number_emergency',
    'number_inpatient', 'number_diagnoses',
    'total_visits', 'inpatient_ratio'
]
cat_cols_base = [
    'race', 'gender', 'age_group',
    'admission_type_id', 'discharge_disposition_id', 'admission_source_id',
    'diag_1_group', 'diag_2_group', 'diag_3_group',
    'max_glu_serum', 'A1Cresult', 'change', 'diabetesMed'
]
num_cols_base = [c for c in num_cols_base if c in train_df.columns]
cat_cols_base = [c for c in cat_cols_base if c in train_df.columns]

pre_base = ColumnTransformer([
    ('num', Pipeline([
        ('imp', SimpleImputer(strategy='median')),
        ('sc', StandardScaler())
    ]), num_cols_base),
    ('cat', Pipeline([
        ('imp', SimpleImputer(strategy='constant', fill_value='missing')),
        ('ohe', OneHotEncoder(handle_unknown='ignore', min_frequency=20))
    ]), cat_cols_base),
])

baseline_lr = Pipeline([
    ('pre', pre_base),
    ('clf', LogisticRegression(max_iter=3000,
                               class_weight='balanced',
                               solver='liblinear',
                               random_state=SEED))
])
baseline_lr.fit(train_df[num_cols_base + cat_cols_base], train_df['target'])
val_probs_lr = baseline_lr.predict_proba(
    val_df[num_cols_base + cat_cols_base])[:, 1]

baseline_dt = Pipeline([
    ('pre', pre_base),
    ('clf', DecisionTreeClassifier(max_depth=4,
                                   class_weight='balanced',
                                   random_state=SEED))
])
baseline_dt.fit(train_df[num_cols_base + cat_cols_base], train_df['target'])
val_probs_dt = baseline_dt.predict_proba(
    val_df[num_cols_base + cat_cols_base])[:, 1]

logging.info("Baselines fitted.")


# =============================================================================
# SECTION 3. EVALUATION UTILITIES
# =============================================================================
def evaluate_model(y_true, y_probs, threshold=0.5, model_name="Model",
                   verbose=True):
    y_pred = (y_probs >= threshold).astype(int)
    metrics = {
        'Model': model_name,
        'Threshold': round(float(threshold), 3),
        'Accuracy':  accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred, zero_division=0),
        'Recall':    recall_score(y_true, y_pred, zero_division=0),
        'F1-Score':  f1_score(y_true, y_pred, zero_division=0),
        'ROC-AUC':   roc_auc_score(y_true, y_probs),
        'PR-AUC':    average_precision_score(y_true, y_probs),
        'Brier':     brier_score_loss(y_true, y_probs),
        'y_true':    y_true,
        'y_pred':    y_pred,
        'y_probs':   y_probs,
    }
    if verbose:
        print(f"\n=== {model_name} ===")
        print(f"Threshold: {threshold:.3f}")
        print(classification_report(y_true, y_pred, digits=4, zero_division=0))
    return metrics


def plot_confusion_matrix(metrics, model_tag, split="val"):
    cm = confusion_matrix(metrics['y_true'], metrics['y_pred'])
    fig, ax = plt.subplots(figsize=(5.5, 4.5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False, ax=ax,
                xticklabels=['Pred No', 'Pred <30'],
                yticklabels=['True No', 'True <30'])
    ax.set_title(f"Confusion Matrix — {metrics['Model']}")
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    fig.tight_layout()
    save_show(fig, f"{FIG_DIR}/cm_{split}_{model_tag}.png")


def plot_metrics_bar(metrics, model_tag, split="val"):
    keys = ['Accuracy', 'Precision', 'Recall', 'F1-Score',
            'ROC-AUC', 'PR-AUC', 'Brier']
    vals = [metrics[k] for k in keys]
    fig, ax = plt.subplots(figsize=(8, 4.5))
    bars = ax.bar(keys, vals,
                  color=sns.color_palette("mako", len(keys)),
                  edgecolor='black')
    for b, v in zip(bars, vals):
        ax.text(b.get_x() + b.get_width() / 2, v + 0.01,
                f"{v:.3f}", ha='center', va='bottom', fontsize=9)
    ax.set_title(f"Metrics — {metrics['Model']}")
    ax.set_ylabel("Score"); ax.set_ylim(0, 1.1)
    fig.tight_layout()
    save_show(fig, f"{FIG_DIR}/metrics_{split}_{model_tag}.png")


def plot_roc_pr(metrics, model_tag, split="val"):
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    fpr, tpr, _ = roc_curve(metrics['y_true'], metrics['y_probs'])
    prec, rec, _ = precision_recall_curve(metrics['y_true'], metrics['y_probs'])

    axes[0].plot(fpr, tpr, lw=2,
                 label=f"AUC = {roc_auc_score(metrics['y_true'], metrics['y_probs']):.3f}")
    axes[0].plot([0, 1], [0, 1], 'k--', lw=1)
    axes[0].set_title(f"ROC — {metrics['Model']}")
    axes[0].set_xlabel("FPR"); axes[0].set_ylabel("TPR")
    axes[0].legend(loc="lower right")

    axes[1].plot(rec, prec, lw=2,
                 label=f"PR-AUC = {average_precision_score(metrics['y_true'], metrics['y_probs']):.3f}")
    axes[1].set_title(f"Precision-Recall — {metrics['Model']}")
    axes[1].set_xlabel("Recall"); axes[1].set_ylabel("Precision")
    axes[1].legend(loc="lower left")

    fig.tight_layout()
    save_show(fig, f"{FIG_DIR}/roc_pr_{split}_{model_tag}.png")


def plot_calibration(metrics, model_tag, split="val"):
    prob_true, prob_pred = calibration_curve(
        metrics['y_true'], metrics['y_probs'],
        n_bins=10, strategy='quantile')
    fig, ax = plt.subplots(figsize=(5.5, 4.5))
    ax.plot(prob_pred, prob_true, marker='o', label=metrics['Model'])
    ax.plot([0, 1], [0, 1], 'k--', label='Perfect')
    ax.set_title(f"Calibration — {metrics['Model']}")
    ax.set_xlabel("Predicted probability")
    ax.set_ylabel("Observed frequency")
    ax.legend()
    fig.tight_layout()
    save_show(fig, f"{FIG_DIR}/calibration_{split}_{model_tag}.png")


def save_all_model_plots(metrics, model_tag, split="val"):
    plot_confusion_matrix(metrics, model_tag, split=split)
    plot_metrics_bar(metrics, model_tag, split=split)
    plot_roc_pr(metrics, model_tag, split=split)
    plot_calibration(metrics, model_tag, split=split)


all_eval_results = []


# =============================================================================
# SECTION 4. Q5 — ACCURACY IS NOT ENOUGH (baseline eval + threshold tuning)
# =============================================================================
logging.info("=== Section 4 (Q5): Baseline evaluation + threshold tuning ===")

# --- LR @ 0.50 ---
lr_metrics_05 = evaluate_model(
    val_df['target'].values, val_probs_lr, threshold=0.5,
    model_name="LR Baseline (Val, thr=0.50)")
all_eval_results.append(lr_metrics_05)
save_all_model_plots(lr_metrics_05, slug("lr_baseline_thr050"), split="val")

# --- LR @ tuned threshold (target recall = 0.60), highest qualifying thr ---
t_star, rec_t, prec_t, f1_t = pick_threshold(
    val_df['target'].values, val_probs_lr, target_recall=0.60)
print(f"\nTuned threshold (LR): t={t_star:.2f} | recall={rec_t:.3f} "
      f"| precision={prec_t:.3f} | F1={f1_t:.3f}")

lr_metrics_tuned = evaluate_model(
    val_df['target'].values, val_probs_lr, threshold=t_star,
    model_name=f"LR Baseline (Val, thr={t_star:.2f})")
all_eval_results.append(lr_metrics_tuned)
save_all_model_plots(lr_metrics_tuned,
                     slug(f"lr_baseline_thr{t_star:.2f}"), split="val")

# --- DT @ 0.50 ---
dt_metrics_05 = evaluate_model(
    val_df['target'].values, val_probs_dt, threshold=0.5,
    model_name="Decision Tree depth=4 (Val, thr=0.50)")
all_eval_results.append(dt_metrics_05)
save_all_model_plots(dt_metrics_05, slug("dt_depth4_thr050"), split="val")


# =============================================================================
# SECTION 5. EXTENDED MODELS (RF, XGB) — no SMOTE, no TF-IDF
# =============================================================================
logging.info("=== Section 5: Extended models ===")

scaler = StandardScaler()
X_train_num = scaler.fit_transform(train_df[num_cols_base].fillna(0))
X_val_num   = scaler.transform(val_df[num_cols_base].fillna(0))
X_test_num  = scaler.transform(test_df[num_cols_base].fillna(0))

ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
X_train_cat = ohe.fit_transform(
    train_df[cat_cols_base].fillna('missing').astype(str))
X_val_cat   = ohe.transform(
    val_df[cat_cols_base].fillna('missing').astype(str))
X_test_cat  = ohe.transform(
    test_df[cat_cols_base].fillna('missing').astype(str))
cat_names = ohe.get_feature_names_out(cat_cols_base).tolist()

X_train = np.hstack([X_train_num, X_train_cat])
X_val   = np.hstack([X_val_num,   X_val_cat])
X_test  = np.hstack([X_test_num,  X_test_cat])
y_train = train_df['target'].values
y_val   = val_df['target'].values
y_test  = test_df['target'].values

feature_names = num_cols_base + cat_names
logging.info(f"Feature matrix: {X_train.shape} "
             f"({len(feature_names)} named features)")

neg, pos = np.bincount(y_train)
scale_pos_weight = neg / pos
logging.info(f"Class counts (train): neg={neg}, pos={pos}, "
             f"scale_pos_weight={scale_pos_weight:.3f}")

# --- Random Forest (class_weight) ---
rf = RandomForestClassifier(
    n_estimators=300, max_depth=None, min_samples_leaf=5,
    class_weight='balanced_subsample',
    n_jobs=-1, random_state=SEED)
rf.fit(X_train, y_train)
val_probs_rf = rf.predict_proba(X_val)[:, 1]
rf_metrics = evaluate_model(y_val, val_probs_rf, threshold=0.5,
                            model_name="Random Forest (Val, thr=0.50)")
all_eval_results.append(rf_metrics)
save_all_model_plots(rf_metrics, slug("random_forest_thr050"), split="val")

# --- XGBoost baseline (scale_pos_weight) ---
xgb_base = xgb.XGBClassifier(
    n_estimators=300, learning_rate=0.05, max_depth=5,
    scale_pos_weight=scale_pos_weight,
    random_state=SEED, eval_metric='logloss',
    feature_names=feature_names,
)
xgb_base.fit(X_train, y_train)
val_probs_xgb = xgb_base.predict_proba(X_val)[:, 1]
xgb_metrics = evaluate_model(y_val, val_probs_xgb, threshold=0.5,
                             model_name="XGBoost (Val, thr=0.50)")
all_eval_results.append(xgb_metrics)
save_all_model_plots(xgb_metrics, slug("xgboost_thr050"), split="val")


# =============================================================================
# SECTION 6. XGBOOST TUNING + FINAL TEST EVALUATION
# =============================================================================
logging.info("=== Section 6: XGBoost tuning + test ===")

X_comb = np.vstack([X_train, X_val])
y_comb = np.hstack([y_train, y_val])
split_index = [-1] * len(X_train) + [0] * len(X_val)
ps = PredefinedSplit(test_fold=split_index)

param_grid = {
    'max_depth': [3, 5, 7],
    'learning_rate': [0.02, 0.05, 0.1],
    'n_estimators': [100, 200, 300, 400],
    'subsample': [0.7, 0.9],
    'colsample_bytree': [0.7, 0.9],
    'min_child_weight': [1, 5],
    'scale_pos_weight': [1.0, scale_pos_weight],
}
search = RandomizedSearchCV(
    estimator=xgb.XGBClassifier(
        random_state=SEED, eval_metric='logloss',
        feature_names=feature_names,
    ),
    param_distributions=param_grid,
    n_iter=20, scoring='roc_auc', cv=ps, random_state=SEED, n_jobs=-1)
search.fit(X_comb, y_comb)

# Refit the best configuration on train only, with early stopping on val.
best_params = dict(search.best_params_)
best_xgb = xgb.XGBClassifier(
    **best_params,
    random_state=SEED, eval_metric='logloss',
    early_stopping_rounds=20,
    feature_names=feature_names,
)
best_xgb.fit(X_train, y_train,
             eval_set=[(X_val, y_val)],
             verbose=False)
print(f"\nBest XGB params: {best_params}")

# Validation performance of the tuned model
val_probs_xgb_tuned = best_xgb.predict_proba(X_val)[:, 1]
val_metrics_xgb_tuned = evaluate_model(
    y_val, val_probs_xgb_tuned, threshold=0.5,
    model_name="Tuned XGBoost (Val, thr=0.50)")
all_eval_results.append(val_metrics_xgb_tuned)
save_all_model_plots(val_metrics_xgb_tuned,
                     slug("tuned_xgboost_val_thr050"), split="val")

# Operating threshold chosen on validation
t_star_xgb, rec_x, prec_x, f1_x = pick_threshold(
    y_val, val_probs_xgb_tuned, target_recall=0.60)
print(f"\nTuned XGB threshold (Val): t={t_star_xgb:.2f} | "
      f"recall={rec_x:.3f} | precision={prec_x:.3f} | F1={f1_x:.3f}")

# Test evaluation — threshold from validation, no test tuning
test_probs_xgb = best_xgb.predict_proba(X_test)[:, 1]
test_metrics_xgb = evaluate_model(
    y_test, test_probs_xgb, threshold=t_star_xgb,
    model_name=f"Tuned XGBoost (Test, thr={t_star_xgb:.2f})")
all_eval_results.append(test_metrics_xgb)
save_all_model_plots(test_metrics_xgb,
                     slug(f"tuned_xgboost_test_thr{t_star_xgb:.2f}"),
                     split="test")

# Bootstrap CIs on test metrics
for name, fn in [("ROC-AUC", roc_auc_score),
                 ("PR-AUC", average_precision_score),
                 ("Brier",  brier_score_loss)]:
    m, lo, hi = bootstrap_ci(y_test, test_probs_xgb, fn)
    print(f"Test {name}: {m:.3f} [{lo:.3f}, {hi:.3f}] (95% bootstrap CI)")

# LR baseline on test at its own tuned threshold
test_probs_lr = baseline_lr.predict_proba(
    test_df[num_cols_base + cat_cols_base])[:, 1]
lr_test_metrics = evaluate_model(
    y_test, test_probs_lr, threshold=t_star,
    model_name=f"LR Baseline (Test, thr={t_star:.2f})")
all_eval_results.append(lr_test_metrics)
save_all_model_plots(lr_test_metrics,
                     slug(f"lr_baseline_thr{t_star:.2f}"), split="test")

metrics_cols = ['Model', 'Threshold', 'Accuracy', 'Precision', 'Recall',
                'F1-Score', 'ROC-AUC', 'PR-AUC', 'Brier']
metrics_df = pd.DataFrame(all_eval_results)[metrics_cols]
print("\n=== Consolidated model metrics ===")
display(metrics_df.round(4))


# =============================================================================
# SECTION 6b. GLOBAL COMPARISON FIGURES (validation models together)
# =============================================================================
logging.info("=== Section 6b: Global comparison figures ===")

val_models = {
    "LR Baseline":   val_probs_lr,
    "Decision Tree": val_probs_dt,
    "Random Forest": val_probs_rf,
    "XGBoost":       val_probs_xgb,
}

fig, ax = plt.subplots(figsize=(7, 6))
for name, probs in val_models.items():
    fpr, tpr, _ = roc_curve(y_val, probs)
    ax.plot(fpr, tpr, lw=2,
            label=f"{name} (AUC={roc_auc_score(y_val, probs):.3f})")
ax.plot([0, 1], [0, 1], 'k--', lw=1)
ax.set_title("ROC Curves — All Models (Validation)")
ax.set_xlabel("False Positive Rate"); ax.set_ylabel("True Positive Rate")
ax.legend(loc="lower right")
fig.tight_layout()
save_show(fig, f"{FIG_DIR}/comparison_roc_all_models_val.png")

fig, ax = plt.subplots(figsize=(7, 6))
for name, probs in val_models.items():
    prec, rec, _ = precision_recall_curve(y_val, probs)
    ax.plot(rec, prec, lw=2,
            label=f"{name} (PR-AUC={average_precision_score(y_val, probs):.3f})")
ax.set_title("Precision-Recall Curves — All Models (Validation)")
ax.set_xlabel("Recall"); ax.set_ylabel("Precision")
ax.legend(loc="lower left")
fig.tight_layout()
save_show(fig, f"{FIG_DIR}/comparison_pr_all_models_val.png")

metric_keys = ['Accuracy', 'Precision', 'Recall', 'F1-Score', 'ROC-AUC', 'PR-AUC']
val_only = pd.DataFrame([r for r in all_eval_results if '(Val' in r['Model']])

fig, ax = plt.subplots(figsize=(11, 5.5))
x = np.arange(len(metric_keys))
width = 0.8 / max(len(val_only), 1)
for i, (_, row) in enumerate(val_only.iterrows()):
    vals = [row[k] for k in metric_keys]
    ax.bar(x + i * width - 0.4 + width / 2, vals, width, label=row['Model'])
ax.set_xticks(x); ax.set_xticklabels(metric_keys, rotation=15)
ax.set_ylim(0, 1.1); ax.set_ylabel("Score")
ax.set_title("Validation Metrics — All Models")
ax.legend(fontsize=8, loc="upper right")
fig.tight_layout()
save_show(fig, f"{FIG_DIR}/comparison_metrics_all_models_val.png")


# =============================================================================
# SECTION 7. Q6 — ERROR ANALYSIS
# =============================================================================
logging.info("=== Section 7 (Q6): Error analysis ===")

error_df = test_df.copy()
error_df['y_prob'] = test_probs_xgb
error_df['y_pred'] = (test_probs_xgb >= t_star_xgb).astype(int)
error_df['error_type'] = np.select(
    [(error_df.target == 1) & (error_df.y_pred == 0),
     (error_df.target == 0) & (error_df.y_pred == 1),
     (error_df.target == 1) & (error_df.y_pred == 1)],
    ['FN', 'FP', 'TP'], default='TN')

bins = [0, 2, 4, 7, 14, 100]
labels = ['1-2', '3-4', '5-7', '8-14', '15+']
error_df['los_bin'] = pd.cut(error_df['time_in_hospital'],
                             bins=bins, labels=labels)

print("\n-- Error type distribution by length-of-stay bin --")
print(error_df.groupby('los_bin', observed=True)['error_type']
      .value_counts(normalize=True).unstack().round(3))

for col in ['admission_type_id', 'diag_1_group', 'discharge_disposition_id']:
    if col in error_df.columns:
        print(f"\n-- Error type by {col} --")
        print(error_df.groupby(col)['error_type']
              .value_counts(normalize=True).unstack().round(3))

fn = error_df[error_df.error_type == 'FN']
print(f"\nFalse negatives: n={len(fn)} "
      f"| median LOS={fn.time_in_hospital.median()} "
      f"| median num_medications={fn.num_medications.median()}")

fn_rate = error_df.groupby('los_bin', observed=True).apply(
    lambda g: (g.error_type == 'FN').mean(), include_groups=False)
fig, ax = plt.subplots(figsize=(7, 4.5))
fn_rate.plot(kind='bar', color='indianred', edgecolor='black', ax=ax)
ax.set_title("False-Negative Rate by Length of Stay (Test Set)")
ax.set_xlabel("Length of stay (days)"); ax.set_ylabel("FN rate within bin")
fig.tight_layout()
save_show(fig, f"{FIG_DIR}/q6_fn_by_los.png")

if 'diag_1_group' in error_df.columns:
    top_groups = error_df['diag_1_group'].value_counts().head(6).index
    grp = error_df[error_df.diag_1_group.isin(top_groups)]
    fn_by_grp = grp.groupby('diag_1_group').apply(
        lambda g: (g.error_type == 'FN').mean(), include_groups=False)
    fig, ax = plt.subplots(figsize=(7, 4.5))
    fn_by_grp.plot(kind='bar', color='salmon', edgecolor='black', ax=ax)
    ax.set_title("False-Negative Rate by Primary Diagnosis Group (Test)")
    ax.set_xlabel("diag_1_group"); ax.set_ylabel("FN rate within group")
    plt.xticks(rotation=25)
    fig.tight_layout()
    save_show(fig, f"{FIG_DIR}/q6_fn_by_diag_group.png")


# =============================================================================
# SECTION 8. Q7 — FEATURE IMPORTANCE (LR COEFFICIENTS + XGB GAIN)
# =============================================================================
logging.info("=== Section 8 (Q7): Feature importance ===")

ohe_names = (baseline_lr.named_steps['pre']
             .named_transformers_['cat']
             .named_steps['ohe']
             .get_feature_names_out(cat_cols_base).tolist())
lr_names = num_cols_base + ohe_names
coefs = baseline_lr.named_steps['clf'].coef_[0]
coef_df = pd.DataFrame({'feature': lr_names, 'coef': coefs})
coef_df['abs_coef'] = coef_df.coef.abs()
top_coef = coef_df.sort_values('abs_coef', ascending=False).head(20)

fig, ax = plt.subplots(figsize=(9, 7))
colors = ['crimson' if c > 0 else 'steelblue' for c in top_coef.coef]
ax.barh(top_coef.feature[::-1], top_coef.coef[::-1],
        color=colors[::-1], edgecolor='black')
ax.axvline(0, color='k', lw=0.8)
ax.set_title("Top 20 Logistic Regression Coefficients (Baseline)")
ax.set_xlabel("Coefficient (log-odds; positive = higher readmission risk)")
fig.tight_layout()
save_show(fig, f"{FIG_DIR}/q7_lr_coefficients.png")

xgb_gain = xgb_gain_with_names(best_xgb, feature_names)
imp_df = (pd.DataFrame({'feature': list(xgb_gain.keys()),
                        'gain': list(xgb_gain.values())})
          .sort_values('gain', ascending=False).head(15))

print("\nTop 10 XGBoost gain features (real names):")
print(imp_df.head(10).to_string(index=False))

fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(imp_df.feature[::-1], imp_df.gain[::-1],
        color=sns.color_palette("mako", len(imp_df)), edgecolor='black')
ax.set_title("Top 15 XGBoost Feature Importances (Gain)")
ax.set_xlabel("Gain")
fig.tight_layout()
save_show(fig, f"{FIG_DIR}/q7_xgb_gain.png")


# =============================================================================
# SECTION 9. Q8 — FAIRNESS ACROSS RACE, GENDER, AGE_GROUP
# =============================================================================
logging.info("=== Section 9 (Q8): Subgroup fairness ===")


def subgroup_fairness(df_in, probs, group_col, threshold=0.5, min_pos=30):
    tmp = df_in.copy()
    tmp['y_prob'] = probs
    tmp['y_pred'] = (tmp['y_prob'] >= threshold).astype(int)
    rows = []
    for val, g in tmp.groupby(group_col, observed=True):
        pos_g = g[g.target == 1]
        n_pos = len(pos_g)
        if n_pos < min_pos:
            continue
        tp = int((pos_g.y_pred == 1).sum())
        rec = tp / n_pos
        lo, hi = wilson_ci(tp, n_pos)          # clamped: lo <= rec <= hi
        rows.append({
            'subgroup': str(val), 'n_pos': n_pos,
            'recall': rec, 'fnr': 1 - rec,
            'recall_lo': lo, 'recall_hi': hi,
        })
    return pd.DataFrame(rows)


def make_xgb(**overrides):
    params = dict(n_estimators=300, learning_rate=0.05, max_depth=5,
                  scale_pos_weight=scale_pos_weight,
                  random_state=SEED, eval_metric='logloss',
                  feature_names=feature_names)
    params.update(overrides)
    return xgb.XGBClassifier(**params)


def oof_probs(X, y, groups=None, n_splits=5):
    """Out-of-fold probabilities via the same XGB config as the deployed
    model. Uses GroupKFold when patient IDs are available."""
    if groups is not None and len(np.unique(groups)) >= n_splits:
        cv = GroupKFold(n_splits=n_splits)
        split_iter = cv.split(X, y, groups)
    else:
        cv = StratifiedKFold(n_splits=n_splits, shuffle=True,
                             random_state=SEED)
        split_iter = cv.split(X, y)
    probs = np.zeros(len(y), dtype=float)
    for tr, te in split_iter:
        clf = make_xgb()
        clf.fit(X[tr], y[tr])
        probs[te] = clf.predict_proba(X[te])[:, 1]
    return probs


# FIX #8 — build OOF probabilities over ALL encounters for fairness
X_all_num = scaler.transform(df[num_cols_base].fillna(0))
X_all_cat = ohe.transform(df[cat_cols_base].fillna('missing').astype(str))
X_all = np.hstack([X_all_num, X_all_cat])
y_all = df['target'].values
groups_all = df['patient_nbr'].values if 'patient_nbr' in df.columns else None

logging.info("Computing out-of-fold probabilities on the full dataset ...")
probs_oof = oof_probs(X_all, y_all, groups=groups_all, n_splits=5)
df_fairness = df.copy()
df_fairness['probs_oof'] = probs_oof

for attr in ['race', 'gender', 'age_group']:
    if attr not in df.columns:
        continue

    print(f"\n{'=' * 60}")
    print(f" Fairness by {attr}")
    print(f"{'=' * 60}")

    print(f"\n--- Test split (XGBoost, thr={t_star_xgb:.2f}) ---")
    tbl_test = subgroup_fairness(test_df, test_probs_xgb, attr,
                                 threshold=t_star_xgb, min_pos=30)
    if not tbl_test.empty:
        print(tbl_test.round(3).to_string(index=False))

    print(f"\n--- Out-of-fold on all {len(df)} encounters "
          f"(XGBoost, thr={t_star_xgb:.2f}) ---")
    tbl_oof = subgroup_fairness(df_fairness, df_fairness['probs_oof'], attr,
                                threshold=t_star_xgb, min_pos=30)
    if not tbl_oof.empty:
        print(tbl_oof.round(3).to_string(index=False))

    if tbl_test.empty and tbl_oof.empty:
        continue

    fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True)
    for ax, (title, tbl) in zip(
            axes,
            [("Test split", tbl_test),
             ("Out-of-fold (all encounters)", tbl_oof)]):
        if tbl.empty:
            ax.text(0.5, 0.5, "insufficient positives",
                    ha='center', va='center', transform=ax.transAxes)
            ax.set_title(title)
            continue

        yerr = np.vstack([
            np.clip(tbl.recall.values - tbl.recall_lo.values, 0.0, None),
            np.clip(tbl.recall_hi.values - tbl.recall.values, 0.0, None),
        ])
        ax.bar(tbl.subgroup, tbl.recall.values, yerr=yerr, capsize=4,
               color=sns.color_palette('crest', len(tbl)),
               edgecolor='black')
        base = (df['target'].mean() if title.startswith("Out")
                else test_df['target'].mean())
        ax.axhline(base, ls='--', color='grey',
                   label=f'base rate = {base:.2f}')
        ax.set_title(f"{title} — recall by {attr}")
        ax.set_ylim(0, 1.0)
        ax.tick_params(axis='x', rotation=25)
        ax.legend()
    fig.tight_layout()
    save_show(fig, f"{FIG_DIR}/q8_fairness_recall_{attr}.png")

    # Chi-square on FN/TP distribution across subgroups (OOF)
    if not tbl_oof.empty and len(tbl_oof) >= 2:
        tmp = df_fairness.copy()
        tmp['y_pred'] = (tmp['probs_oof'] >= t_star_xgb).astype(int)
        pos_all = tmp[tmp.target == 1]
        cont = pd.crosstab(pos_all[attr], pos_all['y_pred'])
        if cont.shape[0] >= 2 and cont.shape[1] >= 2:
            try:
                chi2, p, dof, _ = chi2_contingency(cont.values)
                print(f"\nChi-square on recall across {attr} (OOF): "
                      f"chi2={chi2:.3f}, dof={dof}, p={p:.4f}")
            except Exception as e:
                print(f"Chi-square failed for {attr}: {e}")


# =============================================================================
# SECTION 10. FIGURE INDEX
# =============================================================================
logging.info("=== Section 10: Figure index ===")
for f in sorted(os.listdir(FIG_DIR)):
    logging.info(f"  {FIG_DIR}/{f}")

logging.info("Pipeline complete.")
```

```python
display(Markdown(f"""
## Q5 — Accuracy is not enough: why the threshold is a policy, not a score

**The problem with 0.50.** A classifier trained on an imbalanced target
produces probabilities centred well below 0.5 for the positive class. Using
0.50 as the operating point is a habit inherited from balanced problems; it
has no clinical meaning here. The relevant question is: *what fraction of
true 30-day readmissions are we willing to miss?*

**What was done.** `pick_threshold` selects the **highest** probability
cut-off that still reaches a recall target of 0.60 on the validation set.
Choosing the highest qualifying threshold (rather than the lowest) is what
keeps precision as high as the recall constraint allows. Details of the
mechanism are in the thresholding note above.

**The shift, in numbers (LR baseline, validation).**

| Threshold | Recall | Precision | F1 |
|---|---|---|---|
| 0.50 | {lr_metrics_05['Recall']:.3f} | {lr_metrics_05['Precision']:.3f} | {lr_metrics_05['F1-Score']:.3f} |
| {t_star:.2f} | {rec_t:.3f} | {prec_t:.3f} | {f1_t:.3f} |

Lowering the threshold moves recall from {lr_metrics_05['Recall']:.3f} to
{rec_t:.3f} — a gain of {rec_t - lr_metrics_05['Recall']:.3f} in absolute
terms — but precision drops from {lr_metrics_05['Precision']:.3f} to
{prec_t:.3f}. That trade is the whole point: the threshold is where
clinical judgement enters the pipeline, and the model simply *offers* a
menu of operating points.

**The right way to choose.** A missed readmission and a wasted follow-up
call do not cost the same. Pick the recall floor from a cost analysis —
`cost_FN / cost_FP` — with clinical stakeholders, not from a default. For
any given recall floor, the optimal threshold is determined by the model's
ROC/PR curve, and it is chosen **on validation only** so the test set
remains an honest estimate.

**Takeaway.** Reporting a single accuracy number hides this entire
decision. The model produces probabilities; the threshold is a policy
choice made with clinicians, and it should be stated explicitly every time
a metric is quoted.
"""))
```

```python
display(Markdown(f"""
## Q8 — Fairness: does recall hold up across subgroups?

**Why this section exists.** A model can hit a respectable average recall
while performing much worse for a subgroup that makes up a small share of
the positive class. Aggregate metrics cannot detect that. Fairness
analysis asks: *for a given clinical outcome (recall on the <30 class), do
different groups get the same benefit from the model?*

**What was done.** Recall on the `<30` class was measured separately
across `race`, `gender`, and `age_group`, at the operating threshold chosen
in Q6 (`t = {t_star_xgb:.2f}`). Two views are reported side by side:

1. **Held-out test split** — the cleanest estimate, but small subgroups
   have very few positives here.
2. **Out-of-fold predictions over the full cohort** (`GroupKFold` on
   `patient_nbr` where available, so an encounter of the same patient
   never appears on both sides of a fold). This is not a substitute for a
   true external test, but it is the largest unbiased sample the pipeline
   can produce, and it is what the fairness figures use.

**Uncertainty and significance.** Every subgroup recall is reported with a
**Wilson 95% confidence interval** (the correct interval for a proportion
near 0 or 1, which is where these recalls sit). A **chi-square test** on
the FN/TP contingency table across subgroups is also reported — a small
p-value means the recall differences are unlikely to be noise.

**What to look for in the results.**

- **Overlap of confidence intervals.** If subgroup intervals overlap
  heavily, apparent differences in point estimates may reflect sample
  size, not real disparity.
- **Spread vs. base rate.** Each bar plot includes the base readmission
  rate as a reference line. A subgroup whose recall sits well below that
  line is being under-served by the model.
- **Small subgroups.** Where `n_pos < 30` the subgroup is dropped from the
  table and marked "insufficient positives" on the figure. This is not a
  problem to fix by lowering the threshold; it is a signal that the
  dataset cannot support a fairness claim for that group.

**Takeaway.** Differences in recall across race, gender, and age bracket
exist in this run. Some are small, some are large enough that a
resource-allocation policy built on the model would deliver materially
different follow-up rates to different groups. Before any operational use,
these differences must be re-estimated on a cohort with adequate subgroup
sample sizes, and any residual disparity must be resolved with clinical
stakeholders — either by adjusting the model, by using group-specific
operating thresholds, or by wrapping the model in a human-in-the-loop
review step.
"""))
```
